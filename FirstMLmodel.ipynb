```python
import pandas as pd
```

```python
df = pd.read_csv('/content/sample_data/carprices.csv')
df
```

```python
dummies = pd.get_dummies(df['Car Model'])
dummies = dummies.astype(int)
dummies
```

```python
merged = pd.concat([df,dummies],axis=1)
merged
```

```python
final = merged.drop(['Car Model'], axis = 1)
final
```

```python
final.rename(columns = {'Mercedez Benz C class':'Mercedez'}, inplace=True)
final
```

```python
final.drop(['Mercedez'], axis = 1, inplace = True)
final
```

```python
X=  final.drop(['Sell Price($)'], axis = 1)
X
```

```python
Y = final['Sell Price($)']
Y
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model.fit(X,Y)
```

```python
model.score(X,Y)
```

```python
model.predict([[45000,4,0,0]])
```

```python
model.predict([[86000,7,0,1]])
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
dfle = df
dfle['Car Model'] = le.fit_transform(dfle['Car Model'])
dfle
```

```python
x = dfle[['Car Model','Mileage','Age(yrs)']].values
x
```

```python
y = dfle['Sell Price($)'].values
y
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
ct = ColumnTransformer([('Car Model', OneHotEncoder(), [0])], remainder = 'passthrough')
```

```python
X = ct.fit_transform(x)
X
```

```python
xx = X[:,1:]
xx
```

```python
model.fit(xx,y)
```

```python
model.predict([[0,1,45000,4]])
```

```python
model.predict([[1,0,86000,7]])
```

```python
model.score(xx,y)
```
